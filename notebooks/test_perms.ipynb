### Composite Permutations

````python
p1 = PermSingleAction([2, 1, 3], name="p1")  # (1 2)
p2 = PermSingleAction([1, 3, 2], name="p2")  # (2 3)
````

````python
chain = PermCompositeAction([p1, p2])
````

````python
chain(2)
````

````python
p3 = PermSingleAction([3, 2, 1], name="p3")  # (1 3)
new_chain = chain * p3
````

````python
new_chain.c
````

````python
p2.c
````

````python
identity = PermSingleAction(list(range(1, 4)), name="e")
chain_with_identity = PermCompositeAction([p1, identity, p2, identity])
````

````python
chain_with_identity.reduced
````

````python
new_chain == p2
````

### 2x2 Rubik's Cube Square (not face) Group

```
   2 ---- 4
  /|     /|
 1 ---- 3 |
 | 6 ---| 8
 |/     |/
 5 ---- 7
```

````python
S8 = PermGroup(8)
T = S8.from_cycle("(1 2 4 3)", name="T")
R = S8.from_cycle("(3 4 8 7)", name="R")
B = S8.from_cycle("(2 4 8 6)", name="B")
````

````python
bsgs = S8.schreier_sims([R, B, T])
````

````python
bsgs
````

````python
bsgs2 = S8.schreier_sims_with_sifting([R, B, T])
````

````python
bsgs2
````

````python
test_idx = 1
bsgs[1][:test_idx], bsgs[1][1]
````

````python
A = PermSingleAction(m=[6,2,4,7,5,3,1,8])
````

````python
A_r_word = S8.word_generation(A, bsgs2)
````

````python
len(A_r_word)
````

````python
A_r_word.reduced_actions
````

````python
(A_r_word * A.inverse()).reduced
````

````python
[len(x) for x in bsgs]
````

````python
[len(x) for x in bsgs2]
````

````python
level_idx = 1
test_idx = 10

accepted_generators = bsgs[level_idx][:test_idx]
candidate = bsgs[level_idx][test_idx]

bsgs_test = S8.schreier_sims(accepted_generators, base=range(level_idx+1,9))
test_word = S8.word_generation(candidate, bsgs_test, base=range(level_idx+1,9))

if test_word is not None:
    print("Redundant:", candidate, candidate.reduced, "Expressed as: ", test_word, test_word.reduced)
else:
    print("Not Redundant:", candidate, candidate.reduced)
````
